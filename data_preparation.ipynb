```python
# Convert object column to datetime
df["dined_at"] = pd.to_datetime(df["dined_at"])

# Create separate columns
df["date_dined"] = df["dined_at"].dt.date
df["time_dined"] = df["dined_at"].dt.time
```

```python
# Convert object column to datetime
df["submitted_at"] = pd.to_datetime(df["submitted_at"])

# Create separate columns
df["date_submitted"] = df["submitted_at"].dt.date
df["time_submitted"] = df["submitted_at"].dt.time
```

```python
def expand_dict_column(df, column_name, prefix=None):
    # Convert string dictionaries to real dictionaries
    parsed = df[column_name].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)
    # Expand into columns
    expanded = parsed.apply(pd.Series)
    # Optional prefix
    if prefix:
        expanded = expanded.add_prefix(prefix)

    return pd.concat([df, expanded], axis=1)
```

```python
df = expand_dict_column(df, "user", prefix="user_")
df = expand_dict_column(df, "rating", prefix="rating_")
```

```python
df['restaurant_id'] = df['restaurant_id'].str.split('/').str[-1]
```
